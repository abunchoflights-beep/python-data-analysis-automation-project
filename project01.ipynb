```python
import yfinance
```

```python
ticker = input("Type the stock code. EG. 'AAPL, NVDA, MSF'")
data = yfinance.download([ticker], period="6mo")
close = data.Close[ticker]
close.plot()
```

## Do the analysis requested by the client/user
* Maximun stock price
* Minimum stock price
* Average stock price

```python
max_price = round(close.max(), 2)
min_price = round(close.min(), 2)
mean_price = round(close.mean(), 2)
```

```python
print(max_price)
print(min_price)
print(mean_price)
```
