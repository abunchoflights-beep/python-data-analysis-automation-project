```python
import yfinance
```

```python
ticker = input("Type the stock code. EG. 'AAPL, NVDA, MSF'")
data = yfinance.download([ticker], period="6mo")
close = data.Close[ticker]
close.plot()
```

## Do the analysis requested by the client/user
* Maximun stock price
* Minimum stock price
* Average stock price

```python
max_price = round(close.max(), 2)
min_price = round(close.min(), 2)
mean_price = round(close.mean(), 2)
```

```python
print(max_price)
print(min_price)
print(mean_price)
```

```python
import pyautogui
import time
from pynput import keyboard

cancel_requested = False

def on_press(key):
    global cancel_requested
    if key == keyboard.Key.esc:
        cancel_requested = True
        print("\n[Canceled] Clicks skipped for this cycle.")
listener = keyboard.Listener(on_press=on_press)
listener.start()

for i in range(12):
    print(f"Loop {i+1}/12: Waiting 18 seconds (press ESC to cancle clicks)")
    cancel_requested = False
    for _ in range(180):
        if cancel_requested:
            break
        time.sleep(0.05)
    if cancel_requested:
        print("Skipping click actions for this round.")
        continue
    time.sleep(16.5)
    current_mouse_x, current_mouse_y = pyautogui.position()
    
    pyautogui.PAUSE = 2.2
    
    pyautogui.click(x=125, y=-1247)
    pyautogui.PAUSE = 0.2
    pyautogui.click(x=125, y=-1247)
    pyautogui.PAUSE = 0.1
    pyautogui.moveTo(current_mouse_x, current_mouse_y)
    
listener.stop()
```
